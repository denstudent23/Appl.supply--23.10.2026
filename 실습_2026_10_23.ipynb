```python
import pandas as pd

# 1. Load CSV files
customers_df = pd.read_csv('customers.csv')
orders_df = pd.read_csv('orders.csv')
order_items_df = pd.read_csv('order_items.csv')
products_df = pd.read_csv('products.csv')

# 2. Merge datasets and calculate total price per item
merged = order_items_df.merge(orders_df, on='order_id').merge(products_df, on='product_id')
merged['total_price'] = merged['quantity'] * merged['price']

# 3. Aggregate metrics per customer
spent_by_customer = merged.groupby('customer_id')['total_price'].sum().reset_index()
orders_by_customer = orders_df.groupby('customer_id').size().reset_index(name='order_count')
items_by_customer = merged.groupby('customer_id')['quantity'].sum().reset_index(name='total_quantity')

# Combine everything into a single summary dataframe
customer_summary = customers_df.merge(spent_by_customer, on='customer_id', how='left') \
                               .merge(orders_by_customer, on='customer_id', how='left') \
                               .merge(items_by_customer, on='customer_id', how='left')

# Fill missing values with 0
customer_summary['total_price'] = customer_summary['total_price'].fillna(0)
customer_summary['order_count'] = customer_summary['order_count'].fillna(0)
customer_summary['total_quantity'] = customer_summary['total_quantity'].fillna(0)

# 4. Extract top customers by different criteria
top_by_spent = customer_summary.sort_values(by='total_price', ascending=False).head(5)
top_by_orders = customer_summary.sort_values(by='order_count', ascending=False).head(5)
top_by_quantity = customer_summary.sort_values(by='total_quantity', ascending=False).head(5)

print("=== Top Customers by Total Spent ===")
print(top_by_spent[['customer_id', 'first_name', 'last_name', 'total_price', 'order_count']].to_string(index=False))

print("\n=== Top Customers by Order Count ===")
print(top_by_orders[['customer_id', 'first_name', 'last_name', 'order_count', 'total_price']].to_string(index=False))

print("\n=== Top Customers by Total Quantity Purchased ===")
print(top_by_quantity[['customer_id', 'first_name', 'last_name', 'total_quantity']].to_string(index=False))
```
